# Why Do Transformers Need Positional Information?

Transformers are incredibly powerful, but they have one major limitation:

> **Transformers do not inherently understand the order of a sequence.**

They see all tokens at once, in parallel, with no built-in notion of:
- who comes first  
- who comes next  
- or how far apart two tokens are  

This is very different from earlier sequence models.

---

## How Earlier Models Handle Order

### **RNNs (Recurrent Neural Networks)**
Process tokens **one at a time**, like reading a sentence left to right.

Order is *built into the architecture*.

### **CNNs (Convolutional Neural Networks)**
Slide filters across positions.  
Local order is *baked in* through spatial structure.

### **Transformers**
Look at **all tokens simultaneously** using self-attention:

- `["the", "cat", "sat"]` → processed in parallel

This gives Transformers great speed and flexibility…  
but they lose the natural sense of **sequence order**.

---

# What Goes Wrong Without Positional Information?

To a Transformer with no positional encoding:

- `"the dog chased the cat"`  
- `"the cat chased the dog"`

contain the same words.  
Attention sees the same set of token embeddings.

It has **no clue** which sentence means what.

Even worse:

The sequences:
- `["the", "cat", "sat"]`  
- `["sat", "cat", "the"]`

produce the same attention interactions, just shuffled.

The model cannot tell:
- who is first  
- who is last  
- how tokens relate in time  
- or which tokens came earlier/later  

This destroys meaning in language.

---

# Why Self-Attention Alone Cannot Infer Order

Self-attention compares tokens based on their **content**, not their **position**.  
It does this by projecting each token into queries (Q), keys (K), and values (V), and then computing similarities between them.

But here’s the critical limitation:

If two sequences contain the **same tokens**, just arranged differently,  
their embeddings — and therefore their Q, K, and V vectors — are the same set of points, simply in a different order.

This means:

- the dot products `QKᵀ` are the same values  
- the attention weights are the same values  
- the model produces the same interactions  
- the meaning of the sequence is lost  

So:

- `"the dog chased the cat"`  
- `"the cat chased the dog"`

look identical to self-attention.

Because nothing in scaled dot-product attention tells the model **who came first** or **which token is next to which**.

In short:

> Self-attention understands **what** each token is,  
> but without positional information, it has no idea **where** each token is.

---

# Why We Must Inject Position

To fix this, we enrich each token embedding with information that tells the model:

- **its absolute position** in the sequence  
- **its relative distance** from other tokens  
- **how far ahead or behind** a token is  

This gives the model a coordinate system.

After adding positional information, attention can answer questions like:

- "Is this the first word?"  
- "Does this verb modify a noun two tokens earlier?"  
- "Is this token near the start?"  
- "How far apart are these two tokens?"  

---

# The Goal of Positional Information

To turn an unordered set of token embeddings into an ordered sequence representation.

We achieve this by adding a **positional embedding** to each token embedding:

final_embedding = token_embedding + positional_info

Now the Transformer knows:
- **what** each token is  
- **where** it is  

This unlocks all higher-level reasoning in the model.
